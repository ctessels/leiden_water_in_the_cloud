# Get Data

```python
from default_parameters import get_parameters

token, device_name_ids = get_parameters()

import pandas as pd

df_sensor_id_name = pd.DataFrame.from_dict(device_name_ids, orient='index', columns=['device_id']).reset_index().set_index('device_id')
df_sensor_id_name.columns=['device_name']

df_sensor_info = pd.read_excel('./data/Water in the cloud (Responses).xlsx')
df_sensor_info.columns = ['timestamp', 'device_name', 'located', 'omschrijving', 'foto_1','foto_2', 
                          'lat_lon','located_datum','located_tijd','foto_3', 'foto_4', 
                          'leeftijd_plant', 'extra_omschrijving', 'soort_plant','located_regio']
df_sensor_info = df_sensor_info[['located','device_name', 'lat_lon','located_datum','located_tijd']]
df_sensor_info['device_name'] = 'firefly2_' + df_sensor_info['device_name'].str[-4:]
df_sensor_info = df_sensor_info.merge(df_sensor_id_name.reset_index(),how='left')
df_sensor_info = df_sensor_info[['located','device_name', 'device_id', 'lat_lon','located_datum','located_tijd']]
df_sensor_info = df_sensor_info.iloc[1:,:].reset_index(drop=True)

df = pd.read_excel('./data/Battery_updated.xlsx')
df.drop_duplicates(inplace=True)
df
```

# Summarize Data

```python
df_max_gateway_time = pd.DataFrame(df.groupby('device')['gateway_receive_time'].max())
df_max_gateway_time.columns = ['max_gateway_receive_time']
df_max_gateway_time.index.name = 'device_id'
df_max_gateway_time = df_max_gateway_time.join(df_sensor_id_name)
df_max_gateway_time.sort_values('max_gateway_receive_time')

active_sensors_info = df_max_gateway_time.reset_index().merge(df_sensor_info)
active_sensors_info = active_sensors_info.merge(df[['gateway_receive_time','device','value']],how='left',left_on=['max_gateway_receive_time','device_id'], right_on = ['gateway_receive_time','device'])
active_sensors_info.to_excel('./data/battery_overview.xlsx', index=False)
active_sensors_info
```
